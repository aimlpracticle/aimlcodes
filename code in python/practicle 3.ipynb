```python
# Rule-Based Expert System
# Using Propositional Logic
# Forward Chaining and Backward Chaining

# ---------------------------------------------------
# 1. KNOWLEDGE BASE
# ---------------------------------------------------

facts = {
    "fever",
    "cough",
    "body_pain"
}

rules = [
    (["fever", "cough"], "flu"),
    (["fever", "body_pain"], "viral_infection"),
    (["cough", "sore_throat"], "cold"),
    (["flu"], "weakness"),
    (["viral_infection"], "weakness"),
    (["cold"], "mild_illness")
]


# ---------------------------------------------------
# 2. FORWARD CHAINING
# ---------------------------------------------------

def forward_chaining(facts, rules):
    known_facts = set(facts)

    print("Initial Facts:")
    print(known_facts)
    print("\nForward Chaining Process:")

    changed = True

    while changed:
        changed = False

        for conditions, conclusion in rules:

            # Check whether all conditions are satisfied
            if all(condition in known_facts for condition in conditions):

                if conclusion not in known_facts:
                    known_facts.add(conclusion)

                    print(
                        "Rule Fired:",
                        conditions,
                        "->",
                        conclusion
                    )

                    changed = True

    return known_facts


# Execute Forward Chaining
result = forward_chaining(facts, rules)

print("\nFinal Derived Facts:")
print(result)
# ---------------------------------------------------
# 3. BACKWARD CHAINING
# ---------------------------------------------------

def backward_chaining(goal, facts, rules, visited=None):

    if visited is None:
        visited = set()

    # If goal is already a known fact
    if goal in facts:
        print("Fact found:", goal)
        return True

    # Avoid infinite loops
    if goal in visited:
        return False

    visited.add(goal)

    print("\nTrying to prove:", goal)

    # Find rules that can produce the goal
    for conditions, conclusion in rules:

        if conclusion == goal:

            print(
                "Using Rule:",
                conditions,
                "->",
                conclusion
            )

            # Prove every condition
            all_proved = True

            for condition in conditions:

                if not backward_chaining(
                    condition,
                    facts,
                    rules,
                    visited
                ):
                    all_proved = False
                    break

            if all_proved:
                print("Goal proved:", goal)
                return True

    print("Cannot prove:", goal)
    return False


# ---------------------------------------------------
# 4. TEST BACKWARD CHAINING
# ---------------------------------------------------

goal = "weakness"

print("Backward Chaining Result:")

if backward_chaining(goal, facts, rules):
    print("\nConclusion:", goal, "is TRUE")
else:
    print("\nConclusion:", goal, "is FALSE")
    # ---------------------------------------------------
# 5. INTERACTIVE EXPERT SYSTEM
# ---------------------------------------------------

def expert_system():

    print("======================================")
    print("   RULE-BASED EXPERT SYSTEM")
    print("======================================")

    print("\nEnter symptoms separated by comma.")
    print("Available symptoms:")
    print("fever, cough, body_pain, sore_throat")

    user_input = input("\nEnter symptoms: ")

    user_facts = set(
        symptom.strip().lower()
        for symptom in user_input.split(",")
        if symptom.strip()
    )

    print("\nYour Facts:")
    print(user_facts)

    print("\nChoose Inference Method:")
    print("1. Forward Chaining")
    print("2. Backward Chaining")

    choice = input("\nEnter choice (1/2): ")

    if choice == "1":

        print("\n--- Forward Chaining ---")

        result = forward_chaining(
            user_facts,
            rules
        )

        print("\nDerived Knowledge:")

        for fact in result:
            print("-", fact)

    elif choice == "2":

        goal = input(
            "\nEnter goal to verify "
            "(flu / viral_infection / cold / weakness): "
        ).strip().lower()

        print("\n--- Backward Chaining ---")

        result = backward_chaining(
            goal,
            user_facts,
            rules
        )

        print("\nFinal Result:")

        if result:
            print(goal, "can be proved.")
        else:
            print(goal, "cannot be proved.")

    else:
        print("Invalid choice!")


# Run Expert System
expert_system()
```
